Описание задачи: 
Создать базовый класс ShippingOption в C#, который будет представлять 
различные опции доставки. На основе этого класса разработать 2-3 производных 
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из 
классов должны быть реализованы новые атрибуты и методы, а также 
переопределены некоторые методы базового класса для демонстрации 
полиморфизма. 
Требования к базовому классу ShippingOption: 
• Атрибуты: ID опции доставки (DeliveryOptionId), Название опции доставки 
(DeliveryOptionName), Стоимость доставки (Cost). 
• Методы: 
o CalculateCost(): метод для расчета стоимости доставки. 
o EstimateDeliveryTime(): метод для оценки времени доставки. 
o GetDeliveryDetails(): метод для получения деталей опции доставки. 
Требования к производным классам: 
1. СтандартнаяДоставка 
(StandardDelivery): 
Должна 
содержать 
дополнительные атрибуты, такие как Среднее время доставки 
(AverageDeliveryTime). 
Метод EstimateDeliveryTime() должен 
переопределен для предоставления среднего времени доставки. 
быть 
2. ЭкспрессДоставка (ExpressDelivery): Должна содержать дополнительные 
атрибуты, такие как Минимальное время доставки (MinDeliveryTime). 
Метод CalculateCost() должен быть переопределен для увеличения 
стоимости доставки в случае необходимости ускорения доставки. 
3. Самовывоз (Pickup) (если требуется третий класс): Должна содержать 
дополнительные атрибуты, такие как Адрес пункта самовывоза 
(PickupAddress). Метод GetDeliveryDetails() должен быть переопределен для 
отображения адреса пункта самовывоза.

```

public interface ITrackableShipping
{
    string TrackingNumber { get; set; }
    void GenerateTrackingNumber();
    string GetTrackingStatus();
}

// Интерфейс для опций доставки, которые могут быть настраиваемыми
public interface ICustomizableShipping
{
    string SpecialInstructions { get; set; }
    void AddSpecialInstructions(string instructions);
    void RemoveSpecialInstructions();
}

// --- Базовый класс: ShippingOption ---
public class ShippingOption : ITrackableShipping
{
    public string DeliveryOptionId { get; set; }
    public string DeliveryOptionName { get; set; }
    public decimal Cost { get; protected set; }

    // Новые атрибуты
    public bool IsTrackable { get; set; }
    public decimal MaxWeightKg { get; set; }
    public bool RequiresSignature { get; set; }
    public string EstimatedRegion { get; set; }

    // Реализация интерфейса ITrackableShipping
    public string TrackingNumber { get; set; }
    public void GenerateTrackingNumber()
    {
        if (IsTrackable)
        {
            TrackingNumber = Guid.NewGuid().ToString().Substring(0, 8).ToUpper();
        }
        else
        {
            TrackingNumber = "N/A";
        }
    }
    public virtual string GetTrackingStatus()
    {
        return IsTrackable ? $"Статус отслеживания для {TrackingNumber}: В пути ({EstimatedRegion})" : "Отслеживание недоступно.";
    }

    public ShippingOption(string id, string name, decimal baseCost,
                          bool isTrackable, decimal maxWeight, bool requiresSignature, string region)
    {
        DeliveryOptionId = id;
        DeliveryOptionName = name;
        Cost = baseCost;
        IsTrackable = isTrackable;
        MaxWeightKg = maxWeight;
        RequiresSignature = requiresSignature;
        EstimatedRegion = region;
        TrackingNumber = "N/A";
    }

    public virtual decimal CalculateCost()
    {
        return Cost;
    }

    public virtual TimeSpan EstimateDeliveryTime()
    {
        return TimeSpan.FromDays(5); // Оценка по умолчанию
    }

    public virtual string GetDeliveryDetails()
    {
        return $"[Опция: {DeliveryOptionName} (ID: {DeliveryOptionId})]\n" +
               $"  Стоимость: {Cost:C}\n" +
               $"  Возможность отслеживания: {(IsTrackable ? "Да" : "Нет")}\n" +
               $"  Максимальный вес: {MaxWeightKg} кг\n" +
               $"  Требуется подпись: {(RequiresSignature ? "Да" : "Нет")}\n" +
               $"  Примерное время доставки: {EstimateDeliveryTime().TotalDays} дн.\n" +
               $"  Регион: {EstimatedRegion}\n" +
               $"  {GetTrackingStatus()}";
    }

    // Новые методы (3, согласно дополнительному условию "3-4")
    public bool CheckWeight(decimal weight)
    {
        return weight <= MaxWeightKg;
    }

    public void SetRequiresSignature(bool required)
    {
        RequiresSignature = required;
    }

    public virtual string GetServiceDescription() // Переименован PrintServiceDescription для возврата строки
    {
        return $"{DeliveryOptionName} предоставляет общую опцию доставки.";
    }
}

// --- Производный класс 1: StandardDelivery (Простое наследование) ---
public class StandardDelivery : ShippingOption
{
    public TimeSpan AverageDeliveryTime { get; set; }

    // Новый атрибут (1, согласно дополнительному условию)
    public bool IsInsured { get; set; }

    public StandardDelivery(string id, string name, decimal baseCost, TimeSpan averageTime,
                            bool isTrackable, decimal maxWeight, bool requiresSignature, string region, bool isInsured)
        : base(id, name, baseCost, isTrackable, maxWeight, requiresSignature, region)
    {
        AverageDeliveryTime = averageTime;
        IsInsured = isInsured;
        IsTrackable = false;
        RequiresSignature = false;
    }

    public override TimeSpan EstimateDeliveryTime()
    {
        return AverageDeliveryTime;
    }

    // Новый метод (1, согласно дополнительному условию)
    public void OfferInsurance(bool offer)
    {
        IsInsured = offer;
        if (IsInsured)
        {
            Cost += Cost * 0.02m;
        }
    }

    public override string GetDeliveryDetails()
    {
        return base.GetDeliveryDetails() +
               $"  Страховка: {(IsInsured ? "Включена" : "Не включена")}\n";
    }

    public override string GetServiceDescription()
    {
        return $"{DeliveryOptionName} предлагает экономичную доставку со средним сроком {AverageDeliveryTime.TotalDays} дн.";
    }
}

// --- Производный класс 2: ExpressDelivery (Простое наследование) ---
public class ExpressDelivery : ShippingOption
{
    public TimeSpan MinDeliveryTime { get; set; }

    // Новый атрибут (1, согласно дополнительному условию)
    public decimal SurchargeRate { get; set; }

    public ExpressDelivery(string id, string name, decimal baseCost, TimeSpan minTime,
                           bool isTrackable, decimal maxWeight, bool requiresSignature, string region, decimal surchargeRate)
        : base(id, name, baseCost, isTrackable, maxWeight, requiresSignature, region)
    {
        MinDeliveryTime = minTime;
        SurchargeRate = surchargeRate;
        IsTrackable = true;
        RequiresSignature = true;
    }

    public override decimal CalculateCost()
    {
        decimal originalCost = base.CalculateCost();
        Cost = originalCost * (1 + SurchargeRate);
        return Cost;
    }

    // Новый метод (1, согласно дополнительному условию)
    public void ApplySurcharge()
    {
        if (SurchargeRate > 0)
        {
            CalculateCost();
        }
    }

    public override TimeSpan EstimateDeliveryTime()
    {
        return MinDeliveryTime;
    }

    public override string GetDeliveryDetails()
    {
        return base.GetDeliveryDetails() +
               $"  Минимальный срок доставки: {MinDeliveryTime.TotalHours} ч\n" +
               $"  Коэффициент наценки за срочность: {SurchargeRate:P}\n";
    }

    public override string GetServiceDescription()
    {
        return $"{DeliveryOptionName} - это быстрая доставка с минимальным сроком {MinDeliveryTime.TotalHours} ч.";
    }
}

// --- Производный класс 3: Pickup (Самовывоз) (Простое наследование) ---
public class Pickup : ShippingOption
{
    public string PickupAddress { get; set; }

    // Новый атрибут (1, согласно дополнительному условию)
    public string OperatingHours { get; set; }

    public Pickup(string id, string name, decimal baseCost, string address,
                  bool isTrackable, decimal maxWeight, bool requiresSignature, string region, string operatingHours)
        : base(id, name, baseCost, isTrackable, maxWeight, requiresSignature, region)
    {
        PickupAddress = address;
        OperatingHours = operatingHours;
        IsTrackable = false;
        RequiresSignature = true;
        Cost = 0;
    }

    public override string GetDeliveryDetails()
    {
        // Для самовывоза полностью переопределяем метод, фокусируясь на деталях самовывоза
        return $"[Опция: {DeliveryOptionName} (ID: {DeliveryOptionId})]\n" +
               $"  Стоимость: {Cost:C}\n" +
               $"  Адрес пункта самовывоза: {PickupAddress}\n" +
               $"  Часы работы: {OperatingHours}\n" +
               $"  Требуется подпись при получении: {(RequiresSignature ? "Да" : "Нет")}\n";
    }

    // Новый метод (1, согласно дополнительному условию)
    public string GetOperatingHours()
    {
        return OperatingHours;
    }

    public override TimeSpan EstimateDeliveryTime()
    {
        return TimeSpan.FromHours(0); // Доступно сразу
    }

    public override decimal CalculateCost()
    {
        return Cost;
    }

    public override string GetServiceDescription()
    {
        return $"{DeliveryOptionName} - это удобный самовывоз по адресу {PickupAddress}.";
    }
}

// --- Класс для демонстрации сложного и множественного наследования ---
// PremiumExpressDelivery наследуется от ExpressDelivery (сложное наследование)
// и реализует ICustomizableShipping (множественное наследование через интерфейс)
public class PremiumExpressDelivery : ExpressDelivery, ICustomizableShipping
{
    // Новые атрибуты (2, для демонстрации сложного наследования)
    public bool DedicatedCourierService { get; set; }
    public string DedicatedCourierName { get; set; }

    // Реализация интерфейса ICustomizableShipping
    public string SpecialInstructions { get; set; } = "Нет специальных инструкций";

    public void AddSpecialInstructions(string instructions)
    {
        SpecialInstructions = instructions;
    }

    public void RemoveSpecialInstructions()
    {
        SpecialInstructions = "Нет специальных инструкций";
    }

    // --- Конструктор ---
    public PremiumExpressDelivery(string id, string name, decimal baseCost, TimeSpan minTime, // <-- minTime здесь!
                                  bool isTrackable, decimal maxWeight, bool requiresSignature, string region, decimal surchargeRate,
                                  bool dedicatedCourier)
        : base(id, name, baseCost, minTime, isTrackable, maxWeight, requiresSignature, region, surchargeRate) // <<< ИСПРАВЛЕНО ЗДЕСЬ
    {
        DedicatedCourierService = dedicatedCourier;
        // Дополнительно увеличиваем наценку для премиум-доставки
        SurchargeRate += 0.05m; // Добавляем еще 5% к существующей наценке экспресс-доставки
        Cost = baseCost * (1 + SurchargeRate); // Пересчитываем стоимость с новой наценкой
        IsTrackable = true;         // Всегда отслеживаема
        RequiresSignature = true;   // Всегда требует подписи
        DedicatedCourierName = dedicatedCourier ? "Премиум Курьер A" : "N/A";
    }

    // --- Переопределения методов для PremiumExpressDelivery ---
    public override decimal CalculateCost()
    {
        decimal currentCost = base.CalculateCost();
        if (DedicatedCourierService)
        {
            currentCost += 50m; // Дополнительная фиксированная плата за выделенного курьера
            Cost = currentCost;
        }
        return Cost;
    }

    public override string GetTrackingStatus()
    {
        return base.GetTrackingStatus() + (DedicatedCourierService ? $"\n  -> Ваш заказ обслуживает выделенный курьер: {DedicatedCourierName}." : "");
    }

    // Новый метод (1, для сложного наследования)
    public string ScheduleDedicatedCourier()
    {
        return DedicatedCourierService ? $"Выделенный курьер {DedicatedCourierName} запланирован для доставки." : "Выделенный курьер не предусмотрен для этой услуги.";
    }

    public override TimeSpan EstimateDeliveryTime()
    {
        TimeSpan baseTime = base.EstimateDeliveryTime();
        return TimeSpan.FromHours(baseTime.TotalHours * 0.5); // Уменьшаем вдвое
    }

    public override string GetDeliveryDetails()
    {
        return base.GetDeliveryDetails() +
               $"  Выделенный курьер: {(DedicatedCourierService ? "Да" : "Нет")}\n" +
               $"  Имя курьера: {DedicatedCourierName}\n" +
               $"  Специальные инструкции: {SpecialInstructions}\n";
    }

    public override string GetServiceDescription()
    {
        return $"{DeliveryOptionName} - это высокоприоритетная экспресс-доставка с {(DedicatedCourierService ? "выделенным курьером" : "ускоренным обслуживанием")}.";
    }
}

// --- Основной класс программы для запуска и демонстрации ---

    // Вспомогательный метод для вывода деталей опции доставки
    static void PrintShippingOptionDetails(ShippingOption option)
    {
        Console.WriteLine($"\n--- Детали для опции: {option.DeliveryOptionName} ({option.GetType().Name}) ---");
        Console.WriteLine(option.GetDeliveryDetails());
        decimal calculatedCost = option.CalculateCost();
        Console.WriteLine($"  Рассчитанная стоимость: {calculatedCost:C}");
        TimeSpan estimatedTime = option.EstimateDeliveryTime();
        Console.WriteLine($"  Оценочное время доставки: {estimatedTime.TotalDays} дн.");
        bool weightCheck = option.CheckWeight(15m);
        Console.WriteLine($"  Проверка веса (15 кг): {(weightCheck ? "В пределах" : $"Превышает лимит {option.MaxWeightKg} кг")}");
        option.GenerateTrackingNumber();
        Console.WriteLine($"  {option.GetTrackingStatus()}");
        Console.WriteLine($"  Описание услуги: {option.GetServiceDescription()}");
    }

    
    
        Console.OutputEncoding = System.Text.Encoding.UTF8; // Обеспечивает корректное отображение кириллицы

        Console.WriteLine("--- Демонстрация простого, сложного и множественного наследования в C# ---");
        Console.WriteLine("\n=======================================================\n");

        // 1. Создание и тестирование: Стандартная Доставка (Простое наследование)
        Console.WriteLine(">>> Создание: Стандартная Доставка");
        StandardDelivery standard = new StandardDelivery(
            "STD001", "Стандартная Доставка", 10.00m, TimeSpan.FromDays(3),
            false, 20m, false, "Центральный", true
        );
        PrintShippingOptionDetails(standard);
        Console.WriteLine($"  Добавление страховки:");
        standard.OfferInsurance(true); // Демонстрация нового метода
        Console.WriteLine($"    Новая стоимость со страховкой: {standard.Cost:C}");
        Console.WriteLine("\n=======================================================\n");

        // 2. Создание и тестирование: Экспресс Доставка (Простое наследование)
        Console.WriteLine(">>> Создание: Экспресс Доставка");
        ExpressDelivery express = new ExpressDelivery(
            "EXP001", "Экспресс Доставка", 25.00m, TimeSpan.FromHours(24),
            true, 10m, true, "Северный", 0.10m
        );
        PrintShippingOptionDetails(express);
        Console.WriteLine($"  Применение наценки за срочность:");
        express.ApplySurcharge(); // Демонстрация нового метода
        Console.WriteLine($"    Новая стоимость с наценкой: {express.Cost:C}");
        Console.WriteLine("\n=======================================================\n");

        // 3. Создание и тестирование: Самовывоз (Простое наследование)
        Console.WriteLine(">>> Создание: Самовывоз");
        Pickup pickup = new Pickup(
            "PCK001", "Самовывоз из Пункта A", 0.00m, "Улица Пушкина, 10, Город N",
            false, 50m, true, "Южный", "Пн-Пт 9:00-18:00"
        );
        PrintShippingOptionDetails(pickup);
        Console.WriteLine($"  Часы работы пункта самовывоза: {pickup.GetOperatingHours()}"); // Демонстрация нового метода
        Console.WriteLine("\n=======================================================\n");

        // 4. Создание и тестирование: Премиум Экспресс Доставка (Сложное и Множественное наследование)
        Console.WriteLine(">>> Создание: Премиум Экспресс Доставка");
        PremiumExpressDelivery premiumExpress = new PremiumExpressDelivery(
            "PEXP001", "Премиум Экспресс Доставка", 50.00m, TimeSpan.FromHours(12),
            true, 5m, true, "Восточный", 0.15m, true
        );
        PrintShippingOptionDetails(premiumExpress);
        Console.WriteLine($"  {premiumExpress.ScheduleDedicatedCourier()}"); // Демонстрация нового метода
        Console.WriteLine($"  Добавление спец. инструкций:");
        premiumExpress.AddSpecialInstructions("Позвонить за 30 минут до доставки."); // Демонстрация метода интерфейса ICustomizableShipping
        Console.WriteLine($"    Текущие спец. инструкции: {premiumExpress.SpecialInstructions}");
        Console.WriteLine($"  Удаление спец. инструкций:");
        premiumExpress.RemoveSpecialInstructions(); // Демонстрация метода интерфейса ICustomizableShipping
        Console.WriteLine($"    Текущие спец. инструкции: {premiumExpress.SpecialInstructions}");
        Console.WriteLine("\n=======================================================\n");

        // --- Демонстрация полиморфизма (использование базового типа ShippingOption) ---
        Console.WriteLine("--- Демонстрация полиморфизма (использование базового типа ShippingOption) ---");
        List<ShippingOption> options = new List<ShippingOption>
        {
            standard,
            express,
            pickup,
            premiumExpress // Производные классы могут быть сохранены как базовый тип
        };

        foreach (var opt in options)
        {
            PrintShippingOptionDetails(opt); // Единый метод для вывода деталей
        }

        Console.WriteLine("\n--- Демонстрация использования интерфейсов (множественное наследование) ---");

        // Доступ к методам ITrackableShipping через интерфейс
        ITrackableShipping trackableOption = premiumExpress; // PremiumExpressDelivery реализует ITrackableShipping
        Console.WriteLine($"\n>>> Работа с ITrackableShipping через PremiumExpressDelivery:");
        trackableOption.GenerateTrackingNumber();
        Console.WriteLine($"  Статус отслеживания через интерфейс: {trackableOption.GetTrackingStatus()}");

        // Доступ к методам ICustomizableShipping через интерфейс
        ICustomizableShipping customizableOption = premiumExpress; // PremiumExpressDelivery реализует ICustomizableShipping
        Console.WriteLine($"\n>>> Работа с ICustomizableShipping через PremiumExpressDelivery:");
        customizableOption.AddSpecialInstructions("Оставить у двери соседа, если никого нет дома.");
        Console.WriteLine($"  Специальные инструкции через интерфейс: {customizableOption.SpecialInstructions}");
        customizableOption.RemoveSpecialInstructions();
        Console.WriteLine($"  Специальные инструкции после удаления: {customizableOption.SpecialInstructions}");
    

```
