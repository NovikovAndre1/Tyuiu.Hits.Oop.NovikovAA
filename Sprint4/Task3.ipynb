```
public interface ILogger
{
    void Log(string message);
    void LogWarning(string message);
    void LogError(string message);
}

// Конкретная реализация сервиса логирования
public class ConsoleLogger : ILogger
{
    public void Log(string message)
    {
        Console.WriteLine($"[INFO] {DateTime.Now:HH:mm:ss} - {message}");
    }

    public void LogWarning(string message)
    {
        Console.ForegroundColor = ConsoleColor.Yellow;
        Console.WriteLine($"[WARNING] {DateTime.Now:HH:mm:ss} - {message}");
        Console.ResetColor();
    }

    public void LogError(string message)
    {
        Console.ForegroundColor = ConsoleColor.Red;
        Console.WriteLine($"[ERROR] {DateTime.Now:HH:mm:ss} - {message}");
        Console.ResetColor();
    }
}

// Интерфейс для явной реализации (например, для генерации печатной этикетки)
public interface IPrintableLabel
{
    string GenerateShippingLabel();
    string GetBarcodeData(); // Дополнительный метод интерфейса
}

// =======================================================
// Базовый класс: ShippingOption
// =======================================================
public class ShippingOption
{
    // Атрибуты по заданию
    public int DeliveryOptionId { get; set; }
    public string DeliveryOptionName { get; set; }
    public decimal Cost { get; set; } // Базовая стоимость

    // Дополнительные атрибуты (4 штуки)
    public bool IsAvailableForWeekend { get; protected set; } // Доступна ли доставка в выходные?
    public double MaxPackageWeightKg { get; protected set; } // Максимальный вес посылки для этой опции
    public string RequiredPackagingType { get; protected set; } // Тип упаковки (например, "Коробка", "Конверт")
    public decimal InsuranceCostFactor { get; protected set; } // Коэффициент для расчета стоимости страховки

    // Конструктор базового класса
    public ShippingOption(int id, string name, decimal baseCost)
    {
        DeliveryOptionId = id;
        DeliveryOptionName = name;
        Cost = baseCost;
        // Значения по умолчанию для дополнительных атрибутов
        IsAvailableForWeekend = false;
        MaxPackageWeightKg = 30.0;
        RequiredPackagingType = "Стандартная коробка";
        InsuranceCostFactor = 0.01m; // 1% от стоимости для страховки
    }

    // Методы по заданию (виртуальные для переопределения в производных классах)
    public virtual decimal CalculateCost()
    {
        return Cost; // Базовая стоимость без дополнительных расчетов
    }

    public virtual TimeSpan EstimateDeliveryTime()
    {
        return TimeSpan.FromDays(5); // Оценка по умолчанию
    }

    public virtual string GetDeliveryDetails()
    {
        return $"ID: {DeliveryOptionId}, Название: {DeliveryOptionName}, Базовая Стоимость: {Cost:C}";
    }

    // Дополнительные методы (4 штуки)
    // 1. Перегрузка метода для применения промокода (полиморфизм - перегрузка)
    public virtual decimal ApplyPromoCode(string promoCode)
    {
        if (promoCode == "SAVE10")
        {
            Console.WriteLine($"Применяется фиксированная скидка в 10 единиц.");
            return Cost - 10m; // Пример фиксированной скидки
        }
        Console.WriteLine($"Промокод '{promoCode}' не дал фиксированной скидки.");
        return Cost;
    }

    public virtual decimal ApplyPromoCode(string promoCode, decimal currentCost)
    {
        if (promoCode == "DISCOUNT20")
        {
            Console.WriteLine($"Применяется 20% скидка.");
            return currentCost * 0.80m; // Пример процентной скидки
        }
        Console.WriteLine($"Промокод '{promoCode}' не дал процентной скидки.");
        return currentCost;
    }

    // 2. Расчет стоимости страховки
    public decimal GetInsuranceCost(decimal baseCost)
    {
        return baseCost * InsuranceCostFactor;
    }

    // 3. Проверка доступности доставки в выходные
    public bool CheckWeekendDeliveryAvailability()
    {
        return IsAvailableForWeekend;
    }

    // 4. Установка требуемого типа упаковки
    public void SetRequiredPackaging(string packaging)
    {
        RequiredPackagingType = packaging;
        Console.WriteLine($"Тип упаковки обновлен на: {RequiredPackagingType}");
    }
}

// =======================================================
// Производный класс 1: СтандартнаяДоставка (StandardDelivery)
// =======================================================
public class StandardDelivery : ShippingOption, IPrintableLabel // Реализует интерфейс IPrintableLabel
{
    // Дополнительный атрибут по заданию
    public TimeSpan AverageDeliveryTime { get; set; }

    // Дополнительные атрибуты (2 штуки)
    public bool RequiresSignature { get; set; } // Требуется ли подпись при получении
    public double MaxCubicVolumeM3 { get; set; } // Максимальный объем посылки в кубических метрах

    // Конструктор
    public StandardDelivery(int id, string name, decimal baseCost, TimeSpan avgTime)
        : base(id, name, baseCost)
    {
        AverageDeliveryTime = avgTime;
        // Значения для дополнительных атрибутов
        RequiresSignature = false;
        MaxCubicVolumeM3 = 1.0; // 1 куб. метр
        IsAvailableForWeekend = false; // Стандартная доставка обычно не работает в выходные
        MaxPackageWeightKg = 20.0; // Более низкий лимит веса
        RequiredPackagingType = "Стандартная коробка";
    }

    // Переопределение метода EstimateDeliveryTime (полиморфизм - перекрытие)
    public override TimeSpan EstimateDeliveryTime()
    {
        return AverageDeliveryTime;
    }

    // Дополнительные методы (2 штуки)
    // 1. Проверка требования подписи
    public bool CheckSignatureRequirement()
    {
        Console.WriteLine($"Требование подписи для Стандартной Доставки: {RequiresSignature}");
        return RequiresSignature;
    }

    // 2. Проверка вместимости по объему
    public bool CheckVolumeCapacity(double volume)
    {
        bool fits = volume <= MaxCubicVolumeM3;
        Console.WriteLine($"Стандартная Доставка может обработать {volume} м^3: {fits} (Макс: {MaxCubicVolumeM3} м^3)");
        return fits;
    }

    // Переопределение GetDeliveryDetails для включения специфичных деталей Стандартной Доставки
    public override string GetDeliveryDetails()
    {
        return base.GetDeliveryDetails() + $", Среднее время: {AverageDeliveryTime.TotalDays} дней, Требуется подпись: {RequiresSignature}";
    }

    // Явная реализация интерфейса IPrintableLabel
    string IPrintableLabel.GenerateShippingLabel()
    {
        // Только через интерфейс можно вызвать этот метод
        return $"[Стандартная Доставка - Этикетка]\nОтправитель: Склад A\nПолучатель: {DeliveryOptionName}\nID Заказа: {DeliveryOptionId}\nСреднее время: {AverageDeliveryTime.TotalDays} дней\nТребуется подпись: {RequiresSignature}";
    }

    string IPrintableLabel.GetBarcodeData()
    {
        // Только через интерфейс можно вызвать этот метод
        return $"SD{DeliveryOptionId}-{Guid.NewGuid().ToString().Substring(0, 8)}";
    }
}

// =======================================================
// Производный класс 2: ЭкспрессДоставка (ExpressDelivery)
// =======================================================
public class ExpressDelivery : ShippingOption
{
    // Дополнительный атрибут по заданию
    public TimeSpan MinDeliveryTime { get; set; }

    // Дополнительные атрибуты (2 штуки)
    public int PriorityLevel { get; set; } // Уровень приоритета (1 - самый высокий)
    public bool RequiresExtraAcceleration { get; set; } // Нужна ли дополнительная ускоренная доставка

    // Конструктор
    public ExpressDelivery(int id, string name, decimal baseCost, TimeSpan minTime)
        : base(id, name, baseCost)
    {
        MinDeliveryTime = minTime;
        // Значения для дополнительных атрибутов
        PriorityLevel = 1;
        RequiresExtraAcceleration = false; // По умолчанию дополнительное ускорение не требуется
        IsAvailableForWeekend = true; // Экспресс-доставка часто доступна в выходные
        MaxPackageWeightKg = 15.0; // Может быть ниже для более быстрой доставки
        RequiredPackagingType = "Экспресс-пакет";
        InsuranceCostFactor = 0.02m; // Более высокий коэффициент страховки
    }

    // Переопределение метода CalculateCost (полиморфизм - перекрытие)
    // Увеличение стоимости доставки в случае необходимости ускорения доставки
    public override decimal CalculateCost()
    {
        decimal currentCost = base.CalculateCost(); // Получаем базовую стоимость

        // Всегда применяем общий экспресс-премиум
        currentCost *= 1.25m; // 25% наценка за экспресс-сервис

        // Дополнительная стоимость, если "необходимость ускорения доставки" (RequiresExtraAcceleration)
        if (RequiresExtraAcceleration)
        {
            Console.WriteLine("Применяется дополнительная наценка за ускорение экспресс-доставки.");
            currentCost += 10.00m; // Пример фиксированной доплаты за ускорение
        }
        return currentCost;
    }

    // Перегрузка метода CalculateCost для учета веса посылки (полиморфизм - перегрузка)
    public decimal CalculateCost(double packageWeightKg)
    {
        decimal baseCalculatedCost = this.CalculateCost(); // Вызываем переопределенный метод
        if (packageWeightKg > MaxPackageWeightKg)
        {
            Console.WriteLine($"Внимание: Вес посылки {packageWeightKg}кг превышает максимальный {MaxPackageWeightKg}кг для Экспресс-доставки. Применяется дополнительная стоимость.");
            baseCalculatedCost += (decimal)((packageWeightKg - MaxPackageWeightKg) * 0.5); // 0.5 за кг сверх лимита
        }
        return baseCalculatedCost;
    }

    // Переопределение EstimateDeliveryTime для отражения экспресс-характера
    public override TimeSpan EstimateDeliveryTime()
    {
        return MinDeliveryTime;
    }

    // Дополнительные методы (2 штуки)
    // 1. Получение стоимости приоритетного сервиса
    public decimal GetPriorityServiceCost()
    {
        decimal prioritySurcharge = 0;
        if (PriorityLevel == 1) prioritySurcharge = 20m;
        else if (PriorityLevel == 2) prioritySurcharge = 10m;
        Console.WriteLine($"Стоимость приоритетного сервиса уровня {PriorityLevel}: {prioritySurcharge:C}");
        return prioritySurcharge;
    }

    // 2. Проверка гарантированного окна доставки
    public bool IsGuaranteedWithinWindow(TimeSpan currentDeliveryEstimate, TimeSpan guaranteedWindow)
    {
        bool withinWindow = currentDeliveryEstimate <= MinDeliveryTime + guaranteedWindow;
        Console.WriteLine($"Гарантировано в пределах окна: {withinWindow} (Ожидаемое: {currentDeliveryEstimate}, Мин: {MinDeliveryTime}, Окно: {guaranteedWindow})");
        return withinWindow;
    }

    // Переопределение GetDeliveryDetails для включения специфичных деталей Экспресс-Доставки
    public override string GetDeliveryDetails()
    {
        return base.GetDeliveryDetails() + $", Мин. время: {MinDeliveryTime.TotalHours} часов, Приоритет: {PriorityLevel}";
    }
}

// =======================================================
// Производный класс 3: Самовывоз (Pickup)
// =======================================================
public class Pickup : ShippingOption
{
    // Дополнительный атрибут по заданию
    public string PickupAddress { get; set; }

    // Дополнительные атрибуты (2 штуки)
    public string StoreId { get; set; }
    public string OperatingHours { get; set; } // Часы работы (например, "Пн-Пт 9:00 - 18:00")

    // Конструктор
    public Pickup(int id, string name, decimal baseCost, string address)
        : base(id, name, baseCost)
    {
        PickupAddress = address;
        // Значения для дополнительных атрибутов
        StoreId = "STORE001";
        OperatingHours = "Пн-Пт 9:00 - 18:00";
        IsAvailableForWeekend = true; // Многие пункты самовывоза открыты в выходные
        MaxPackageWeightKg = 50.0; // Часто могут обрабатывать более тяжелые товары
        RequiredPackagingType = "Любая подходящая упаковка";
        InsuranceCostFactor = 0.005m; // Ниже риск, ниже страховка
    }

    // Переопределение CalculateCost (Самовывоз часто бесплатный или очень дешевый)
    public override decimal CalculateCost()
    {
        Console.WriteLine("Сервис самовывоза обычно имеет номинальную или нулевую стоимость.");
        return 0m; // Часто бесплатно
    }

    // Переопределение EstimateDeliveryTime (время до готовности к самовывозу)
    public override TimeSpan EstimateDeliveryTime()
    {
        return TimeSpan.FromHours(4); // Готовность через 4 часа
    }

    // Переопределение GetDeliveryDetails (для отображения адреса пункта самовывоза)
    public override string GetDeliveryDetails()
    {
        return base.GetDeliveryDetails() + $", Адрес самовывоза: {PickupAddress}, ID Магазина: {StoreId}";
    }

    // Дополнительные методы (2 штуки)
    // 1. Получение ссылки на карту для проезда
    public string GetStoreDirectionsLink()
    {
        string encodedAddress = Uri.EscapeDataString(PickupAddress);
        string link = $"https://www.google.com/maps/search/?api=1&query={encodedAddress}";
        Console.WriteLine($"Ссылка на карту: {link}");
        return link;
    }

    // 2. Проверка, открыт ли магазин сейчас
    public bool IsStoreOpenNow(DateTime currentTime)
    {
        // Простая имитация, реальная реализация парсила бы OperatingHours
        int hour = currentTime.Hour;
        bool isOpen = (hour >= 9 && hour < 18); // Предполагаем Пн-Пт 9-18 для простоты
        Console.WriteLine($"Магазин открыт в {currentTime.ToShortTimeString()}: {isOpen} (Часы работы: {OperatingHours})");
        return isOpen;
    }
}

// =======================================================
// Generic класс для отображения полной информации о варианте доставки
// (с управлением зависимостями)
// =======================================================
public class DeliveryInfoDisplay<T> where T : ShippingOption
{
    public T DeliveryOption { get; set; }
    private readonly ILogger _logger; // Зависимость от ILogger

    // Конструктор, принимающий зависимость ILogger
    public DeliveryInfoDisplay(T option, ILogger logger)
    {
        DeliveryOption = option;
        _logger = logger; // Инъекция зависимости
    }

    public void DisplayFullDetails()
    {
        _logger.Log($"\n--- Универсальный вывод для '{DeliveryOption.DeliveryOptionName}' (Тип: {DeliveryOption.GetType().Name}) ---");
        _logger.Log($"ID: {DeliveryOption.DeliveryOptionId}");
        _logger.Log($"Базовая Стоимость: {DeliveryOption.Cost:C}");
        _logger.Log($"Рассчитанная Стоимость: {DeliveryOption.CalculateCost():C}"); // Полиморфный вызов
        _logger.Log($"Ожидаемое Время: {DeliveryOption.EstimateDeliveryTime()}"); // Полиморфный вызов
        _logger.Log($"Детали: {DeliveryOption.GetDeliveryDetails()}"); // Полиморфный вызов

        // Вывод некоторых добавленных базовых атрибутов/методов
        _logger.Log($"Доставка в выходные доступна: {DeliveryOption.CheckWeekendDeliveryAvailability()}");
        _logger.Log($"Макс. вес посылки: {DeliveryOption.MaxPackageWeightKg} кг");
        _logger.Log($"Требуемая упаковка: {DeliveryOption.RequiredPackagingType}");
        _logger.Log($"Примерная стоимость страховки: {DeliveryOption.GetInsuranceCost(DeliveryOption.CalculateCost()):C}");

        // Демонстрация использования перегруженных методов ApplyPromoCode
        _logger.Log($"Стоимость после промокода 'SAVE10': {DeliveryOption.ApplyPromoCode("SAVE10"):C}");
        _logger.Log($"Стоимость после промокода 'DISCOUNT20' на текущую стоимость: {DeliveryOption.ApplyPromoCode("DISCOUNT20", DeliveryOption.CalculateCost()):C}");

        // Демонстрация специфичных методов/атрибутов производных классов
        if (DeliveryOption is StandardDelivery sd)
        {
            sd.CheckSignatureRequirement();
            sd.CheckVolumeCapacity(0.5); // Проверка объема 0.5 м^3

            // Демонстрация явной реализации интерфейса IPrintableLabel
            IPrintableLabel printableLabel = sd; // Приведение типа к интерфейсу
            _logger.LogWarning($"--- Демонстрация явной реализации IPrintableLabel ---");
            _logger.Log(printableLabel.GenerateShippingLabel());
            _logger.Log($"Штрих-код: {printableLabel.GetBarcodeData()}");
        }
        else if (DeliveryOption is ExpressDelivery ed)
        {
            ed.GetPriorityServiceCost();
            ed.IsGuaranteedWithinWindow(ed.MinDeliveryTime + TimeSpan.FromMinutes(30), TimeSpan.FromHours(1)); // Проверка гарантии в пределах 1 часа
            _logger.Log($"Стоимость экспресс-доставки для посылки 8кг: {ed.CalculateCost(8.0):C}");
            _logger.Log($"Стоимость экспресс-доставки для посылки 20кг (превышение лимита): {ed.CalculateCost(20.0):C}");
        }
        else if (DeliveryOption is Pickup p)
        {
            p.GetStoreDirectionsLink();
            p.IsStoreOpenNow(DateTime.Now);
        }

        _logger.Log("-------------------------------------------------\n");
    }
}

// =======================================================
// Главный класс программы
// =======================================================

        // Устанавливаем кодировку для корректного отображения символов валют
        Console.OutputEncoding = Encoding.UTF8;

        // Создаем экземпляр сервиса логирования
        ILogger logger = new ConsoleLogger();

        logger.Log("--- Запуск демонстрационной программы ---");

        // 1. Создание экземпляров производных классов
        StandardDelivery standard = new StandardDelivery(101, "Стандартная Доставка", 15.00m, TimeSpan.FromDays(3));
        ExpressDelivery express = new ExpressDelivery(201, "Экспресс Доставка", 25.00m, TimeSpan.FromHours(12));
        express.RequiresExtraAcceleration = true; // Для демонстрации условия "в случае необходимости ускорения доставки"
        Pickup pickup = new Pickup(301, "Самовывоз из Магазина", 5.00m, "ул. Пушкина, д. Колотушкина, 5");

        logger.Log("\n--- Демонстрация Полиморфизма со списком ShippingOption ---");
        // Список базового типа, содержащий объекты производных типов
        List<ShippingOption> options = new List<ShippingOption> { standard, express, pickup };

        foreach (var option in options)
        {
            logger.Log($"\n--- Обработка: {option.DeliveryOptionName} ---");
            logger.Log($"Детали: {option.GetDeliveryDetails()}"); // Полиморфный вызов
            logger.Log($"Рассчитанная Стоимость: {option.CalculateCost():C}"); // Полиморфный вызов
            logger.Log($"Ожидаемое Время: {option.EstimateDeliveryTime()}"); // Полиморфный вызов

            // Демонстрация дополнительных методов базового класса
            logger.Log($"Доставка в выходные доступна: {option.CheckWeekendDeliveryAvailability()}");
            option.SetRequiredPackaging("Надежная коробка"); // Вызовет Console.WriteLine внутри метода
            logger.Log($"Стоимость страховки: {option.GetInsuranceCost(option.CalculateCost()):C}");

            // Демонстрация перегрузки методов базового класса
            decimal costAfterFlatPromo = option.ApplyPromoCode("SAVE10"); // Вызовет Console.WriteLine внутри метода
            logger.Log($"Стоимость после 'SAVE10': {costAfterFlatPromo:C}");
            decimal costAfterPercentagePromo = option.ApplyPromoCode("DISCOUNT20", option.CalculateCost()); // Вызовет Console.WriteLine внутри метода
            logger.Log($"Стоимость после 'DISCOUNT20': {costAfterPercentagePromo:C}");

            // Демонстрация специфичных методов производных классов (с приведением типов)
            if (option is ExpressDelivery exp)
            {
                exp.GetPriorityServiceCost(); // Вызовет Console.WriteLine внутри метода
                exp.IsGuaranteedWithinWindow(exp.MinDeliveryTime + TimeSpan.FromMinutes(30), TimeSpan.FromHours(1)); // Вызовет Console.WriteLine внутри метода
            }
            else if (option is StandardDelivery std)
            {
                std.CheckSignatureRequirement(); // Вызовет Console.WriteLine внутри метода
                // Демонстрация явной реализации интерфейса IPrintableLabel (вне generic класса)
                if (std is IPrintableLabel stdLabel)
                {
                    logger.LogWarning($"--- Демонстрация явной реализации IPrintableLabel (отдельно) ---");
                    logger.Log($"Этикетка для {std.DeliveryOptionName}: \n{stdLabel.GenerateShippingLabel()}");
                    logger.Log($"Штрих-код для {std.DeliveryOptionName}: {stdLabel.GetBarcodeData()}");
                }
            }
            else if (option is Pickup pck)
            {
                pck.GetStoreDirectionsLink(); // Вызовет Console.WriteLine внутри метода
            }
        }

        logger.Log("\n--- Демонстрация Generic Классов с Инъекцией Зависимостей ---");

        // Использование generic класса с разными типами опций доставки,
        // передавая экземпляр ILogger в каждый конструктор (ручная инъекция зависимостей)
        DeliveryInfoDisplay<StandardDelivery> standardDisplay = new DeliveryInfoDisplay<StandardDelivery>(standard, logger);
        standardDisplay.DisplayFullDetails();

        DeliveryInfoDisplay<ExpressDelivery> expressDisplay = new DeliveryInfoDisplay<ExpressDelivery>(express, logger);
        expressDisplay.DisplayFullDetails();

        DeliveryInfoDisplay<Pickup> pickupDisplay = new DeliveryInfoDisplay<Pickup>(pickup, logger);
        pickupDisplay.DisplayFullDetails();

        // Generic класс также может работать с базовым типом
        // (но тогда не будет доступа к специфичным методам производных классов без явного приведения)
        DeliveryInfoDisplay<ShippingOption> genericBaseDisplay = new DeliveryInfoDisplay<ShippingOption>(standard, logger);
        genericBaseDisplay.DisplayFullDetails();

        logger.Log("\n--- Конец Демонстрации ---");
    

```
