<h1 style="color:DodgerBlue">Создание экземпляров классов (объектов) и взаимодействие объектов между собой</h1>

### Создание экземпляров классов (объектов)

В C#, экземпляр класса создается с помощью оператора `new`, который выделяет память для нового объекта и вызывает конструктор класса для инициализации этого объекта. Конструктор — это специальный метод в классе, который имеет то же имя, что и класс, и используется для установки начального состояния объекта.

#### Пример создания экземпляров классов Vehicle и Car

```csharp
Vehicle myVehicle = new Vehicle();
Car myCar = new Car();
```

В этом примере создаются два объекта: один для класса `Vehicle` и другой для класса `Car`. Конструкторы этих классов будут автоматически вызваны при создании объектов.

### Взаимодействие объектов между Собой

Взаимодействие объектов между собой означает, что один объект может вызывать методы другого объекта или обращаться к его свойствам (если они доступны). Это позволяет объектам обмениваться информацией и координировать свои действия.

#### Пример взаимодействия объектов

```csharp
myVehicle.Color = "Red";
myVehicle.Speed = 60;

myCar.Color = "Blue";
myCar.FuelLevel = 100;
myCar.Refuel(50); // Увеличиваем уровень топлива на 50
```

В этом примере мы устанавливаем свойства `Color` и `Speed` для объекта `myVehicle`, а также свойства `Color` и `FuelLevel` для объекта `myCar`. Затем мы вызываем метод `Refuel` у объекта `myCar`, чтобы увеличить его уровень топлива.

### Закрепление правил использования модификаторов доступа

Модификаторы доступа определяют уровень доступности членов класса из других частей кода. Важно правильно использовать модификаторы доступа для обеспечения безопасности и инкапсуляции данных.

#### Примеры модификаторов доступа в классах Vehicle и Car

```csharp
public class Vehicle
{
    public string Color { get; set; } // public свойство
    private int speed; // private поле

    protected void SetSpeed(int value) // protected метод
    {
        speed = value;
    }
}

public class Car : Vehicle
{
    private int fuelLevel; // private поле

    public void Refuel(int amount) // public метод
    {
        fuelLevel += amount;
    }

    protected void CheckSpeedLimit() // protected метод
    {
        if (speed > 120)
        {
            Console.WriteLine("Превышен скоростной режим!");
        }
    }
}
```

В этом примере:
- Свойство `Color` в классе `Vehicle` объявлено как `public`, что позволяет доступ к нему из любого места в коде.
- Поле `speed` в классе `Vehicle` объявлено как `private`, что ограничивает доступ к нему только внутри класса `Vehicle`.
- Метод `SetSpeed` в классе `Vehicle` объявлен как `protected`, что позволяет доступ к нему из класса `Vehicle` и его производных классов.
- Поле `fuelLevel` в классе `Car` объявлено как `private`, ограничивая доступ к нему только внутри класса `Car`.
- Методы `Refuel` и `CheckSpeedLimit` в классе `Car` демонстрируют использование модификаторов доступа для контроля над тем, как объекты могут взаимодействовать друг с другом.

Правильное использование модификаторов доступа обеспечивает безопасность данных и инкапсуляцию, позволяя разработчикам создавать более надежные и управляемые программы.

````
public class Vehicle
{
    // Public свойство Color, доступно отовсюду
    public string Color { get; set; }

    // Protected поле speed. Доступно из Vehicle и из всех классов, унаследованных от Vehicle (например, Car).
    protected int speed;

    // Protected метод SetSpeed. Может быть вызван из Vehicle и из унаследованных классов.
    protected void SetSpeed(int value)
    {
        if (value >= 0) // Простая проверка на не отрицательное значение
        {
            speed = value;
            Console.WriteLine($"Скорость установлена на {speed} км/ч.");
        }
        else
        {
            Console.WriteLine("Скорость не может быть отрицательной!");
        }
    }

    // Конструктор по умолчанию для Vehicle (можно и не писать, он будет создан автоматически)
    public Vehicle()
    {
        Console.WriteLine("Объект Vehicle создан.");
    }

    // Конструктор с параметром
    public Vehicle(string color)
    {
        Color = color;
        Console.WriteLine($"Объект Vehicle создан. Цвет: {Color}.");
    }
}

// Класс Car, наследующийся от Vehicle
public class Car : Vehicle
{
    // Private поле fuelLevel, доступно только внутри класса Car
    private int fuelLevel;

    // Public метод Refuel. Доступен отовсюду.
    public void Refuel(int amount)
    {
        if (amount > 0) // Простая проверка на не отрицательное значение
        {
            fuelLevel += amount;
            Console.WriteLine($"Заправлено {amount} литров топлива. Текущий уровень топлива: {fuelLevel} литров.");
        }
        else
        {
            Console.WriteLine("Нельзя заправить отрицательное количество топлива!");
        }
    }

    // Protected метод CheckSpeedLimit. Может быть вызван из Car и его производных классов.
    // Теперь имеет доступ к полю 'speed' благодаря модификатору 'protected' в Vehicle.
    protected void CheckSpeedLimit()
    {
        if (speed > 120)
        {
            Console.WriteLine($"ВНИМАНИЕ: Превышен скоростной режим! Текущая скорость: {speed} км/ч.");
        }
        else
        {
            Console.WriteLine($"Скорость {speed} км/ч в пределах нормы.");
        }
    }

    // Public метод Drive для демонстрации установки скорости и проверки лимита
    public void Drive(int desiredSpeed)
    {
        Console.WriteLine($"\nПопытка разогнаться до {desiredSpeed} км/ч...");
        SetSpeed(desiredSpeed); // Вызываем protected метод базового класса
        CheckSpeedLimit();      // Вызываем protected метод текущего класса
    }

    // Конструктор для Car
    public Car(string color, int initialFuelLevel = 0) : base(color) // Вызываем конструктор базового класса
    {
        fuelLevel = initialFuelLevel;
        Console.WriteLine($"Объект Car создан. Цвет: {Color}, уровень топлива: {fuelLevel} литров.");
    }
}

// Основной класс для запуска программы


    
        Console.WriteLine("--- Демонстрация работы классов Vehicle и Car ---\n");

        // 1. Создание экземпляра класса Car
        // Используем конструктор Car, который также вызывает конструктор Vehicle
        Car myCar = new Car("Красный", 50);
        Console.WriteLine($"Машина создана. Цвет: {myCar.Color}, текущий уровень топлива: 50 литров."); // Не можем напрямую получить fuelLevel, т.к. private


        // 2. Взаимодействие с объектом Car
        Console.WriteLine("\n--- Взаимодействие с объектом myCar ---");

        // Устанавливаем публичное свойство Color
        myCar.Color = "Синий";
        Console.WriteLine($"Цвет моей машины изменен на: {myCar.Color}");

        // Вызываем публичный метод Refuel
        myCar.Refuel(30);
        myCar.Refuel(10);
        myCar.Refuel(-5); // Попытка заправить отрицательное количество

        // Демонстрируем метод Drive, который устанавливает скорость и проверяет лимит
        myCar.Drive(80);  // Скорость в норме
        myCar.Drive(150); // Превышение скорости
        myCar.Drive(100); // Снова в норме
        myCar.Drive(-20); // Попытка установить отрицательную скорость

        Console.WriteLine("\n--- Демонстрация завершена ---");

        // Для того чтобы консольное окно не закрылось сразу в некоторых средах
        
    
````

````
public class Animal
{
    // Статический атрибут (поле): подсчитывает общее количество созданных животных
    private static int _totalAnimalsCreated = 0;

    // Свойство экземпляра (обычный атрибут): имя животного, только для чтения извне после создания
    public string Name { get; private set; }

    // Полное свойство экземпляра (обычный атрибут) с геттером и сеттером и базовой валидацией
    private int _age;
    public int Age
    {
        get { return _age; }
        set // Сеттер с простой проверкой
        {
            if (value >= 0)
            {
                _age = value;
            }
            else
            {
                Console.WriteLine($"[ПРЕДУПРЕЖДЕНИЕ] Возраст не может быть отрицательным для {Name}.");
                _age = 0;
            }
        }
    }

    // Защищенное свойство: здоровье, доступно Animal и его наследникам.
    // Геттер public, сеттер protected для контроля изнутри иерархии.
    protected int _health;
    public int Health
    {
        get { return _health; }
        protected set // Сеттер с ограничением значения
        {
            _health = Math.Clamp(value, 0, 100); // Ограничиваем здоровье от 0 до 100
        }
    }

    // Конструктор базового класса
    public Animal(string name, int age)
    {
        Name = name;
        Age = age; // Используем сеттер для установки возраста
        Health = 100; // Начальное здоровье
        _totalAnimalsCreated++; // Увеличиваем статический счетчик
        Console.WriteLine($"[Animal] {Name} создан. Всего животных: {_totalAnimalsCreated}");
    }

    // Статический метод: возвращает общее количество созданных животных
    public static int GetTotalAnimalsCount()
    {
        return _totalAnimalsCreated;
    }

    // Обычный (экземплярный) метод: действие "есть"
    public void Eat()
    {
        Console.WriteLine($"{Name} ест.");
        Heal(5); // Немного восстанавливает здоровье
    }

    // Виртуальный метод: издавать звук (будет переопределен в наследниках)
    public virtual void MakeSound()
    {
        Console.WriteLine($"{Name} издает звук.");
    }

    // Защищенный метод: восстановить здоровье, доступен Animal и его наследникам
    protected void Heal(int amount)
    {
        if (Health == 0) return; // Не лечим мертвых
        Console.WriteLine($"{Name} лечится на {amount} очков.");
        Health += amount; // Используем сеттер свойства Health
        Console.WriteLine($"Здоровье {Name}: {Health}/100.");
    }

    // Обычный метод: получить урон
    public void TakeDamage(int amount)
    {
        if (Health == 0) return; // Мертвые не получают урон
        Console.WriteLine($"{Name} получает {amount} урона.");
        Health -= amount; // Используем сеттер свойства Health
        Console.WriteLine($"Здоровье {Name}: {Health}/100.");
        if (Health == 0)
        {
            Console.WriteLine($"{Name} умер.");
        }
    }
}

// --- ПРОИЗВОДНЫЙ КЛАСС: Dog ---
public class Dog : Animal
{
    // Специфический обычный атрибут (свойство) для собаки
    public string Breed { get; set; }

    // Статический атрибут: количество созданных собак
    private static int _totalDogsCreated = 0;
    public static int TotalDogsCreated { get { return _totalDogsCreated; } }

    // Конструктор класса Dog
    public Dog(string name, int age, string breed)
        : base(name, age) // Вызываем конструктор базового класса Animal
    {
        Breed = breed;
        _totalDogsCreated++;
        Console.WriteLine($"[Dog] {Name} (порода: {Breed}) создан. Всего собак: {_totalDogsCreated}");
    }

    // Переопределение метода MakeSound из Animal
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} (собака) говорит: Гав-гав!");
    }

    // Специфический метод для собаки
    public void Fetch(string item)
    {
        Console.WriteLine($"{Name} принес {item}.");
    }

    // Метод взаимодействия с другим животным
    public void BarkAt(Animal otherAnimal)
    {
        Console.WriteLine($"{Name} (собака) лает на {otherAnimal.Name} ({otherAnimal.GetType().Name}).");
        // Пример взаимодействия, где одно животное влияет на другое
        if (otherAnimal is Cat cat) // Если другое животное - кошка
        {
            cat.ReactToDogBark(this); // Кошка реагирует на лай собаки
        }
    }
}

// --- ПРОИЗВОДНЫЙ КЛАСС: Cat ---
public class Cat : Animal
{
    // Специфический обычный атрибут (свойство) для кошки
    public string FurColor { get; set; }

    // Статический атрибут: количество созданных кошек
    private static int _totalCatsCreated = 0;
    public static int TotalCatsCreated { get { return _totalCatsCreated; } }

    // Конструктор класса Cat
    public Cat(string name, int age, string furColor)
        : base(name, age) // Вызываем конструктор базового класса Animal
    {
        FurColor = furColor;
        _totalCatsCreated++;
        Console.WriteLine($"[Cat] {Name} (цвет шерсти: {FurColor}) создан. Всего кошек: {_totalCatsCreated}");
    }

    // Переопределение метода MakeSound из Animal
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} (кошка) говорит: Мяу-мяу!");
    }

    // Специфический метод для кошки
    public void Scratch(string surface)
    {
        Console.WriteLine($"{Name} царапает {surface}.");
    }

    // Метод реакции кошки на собаку (взаимодействие объектов)
    public void ReactToDogBark(Dog barkingDog)
    {
        Console.WriteLine($"{Name} (кошка) шипит на {barkingDog.Name}!");
        TakeDamage(10); // Кошка испытывает стресс от лая
    }

    // Метод взаимодействия с другим животным
    public void Chase(Animal otherAnimal)
    {
        Console.WriteLine($"{Name} (кошка) гоняется за {otherAnimal.Name} ({otherAnimal.GetType().Name}).");
        // Можно добавить дополнительные действия, если необходимо
    }
}


// --- ОСНОВНАЯ ПРОГРАММА (Main) ---


        Console.OutputEncoding = System.Text.Encoding.UTF8; // Для корректного отображения кириллицы

        Console.WriteLine("--- Создание животных ---\n");

        Dog rex = new Dog("Рекс", 5, "Немецкая овчарка");
        Cat whiskers = new Cat("огурец", 3, "Рыжий");
        Dog buddy = new Dog("кирюха", 2, "Золотистый ретривер");
        Cat tom = new Cat("Том", 7, "Серый");

        Console.WriteLine("\n--- Проверка статических атрибутов ---\n");
        Console.WriteLine($"Всего животных создано: {Animal.GetTotalAnimalsCount()}");
        Console.WriteLine($"Всего собак создано: {Dog.TotalDogsCreated}");
        Console.WriteLine($"Всего кошек создано: {Cat.TotalCatsCreated}");

        Console.WriteLine("\n--- Взаимодействие с объектами ---\n");

        // Демонстрация свойств, геттеров и сеттеров
        Console.WriteLine($"Имя Рекса: {rex.Name}, Возраст: {rex.Age}, Здоровье: {rex.Health}");
        rex.Age = 6; // Использование сеттера
        Console.WriteLine($"Новый возраст Рекса: {rex.Age}");
        rex.Age = -1; // Демонстрация валидации в сеттере
        Console.WriteLine($"Возраст Рекса после неверной установки: {rex.Age}");

        Console.WriteLine($"Имя огурца: {whiskers.Name}, Цвет шерсти: {whiskers.FurColor}");
        whiskers.FurColor = "Белый"; // Использование сеттера
        Console.WriteLine($"Новый цвет шерсти огурца: {whiskers.FurColor}");

        // Демонстрация обычных методов
        rex.MakeSound(); // Переопределенный метод
        whiskers.MakeSound(); // Переопределенный метод
        rex.Eat();
        whiskers.Scratch("когтеточку");
        rex.Fetch("палку");

        // Демонстрация взаимодействия между объектами
        Console.WriteLine("\n--- Взаимодействие между Рексом и огурцом ---");
        rex.BarkAt(whiskers); // Собака лает на кошку
        whiskers.Chase(rex);  // Кошка гоняется за собакой

        Console.WriteLine("\n--- Взаимодействие между кирюхой и Томом ---");
        buddy.BarkAt(tom);    // Другая собака лает на другую кошку
        tom.Chase(buddy);     // Другая кошка гоняется за другой собакой

        // Демонстрация здоровья
        Console.WriteLine("\n--- Демонстрация здоровья ---");
        rex.TakeDamage(30);
        rex.Eat(); // Еда восстанавливает здоровье
        rex.TakeDamage(80); // Урон, который должен убить Рекса
    

        Console.WriteLine("\n--- Конец программы ---");
        
    

````
