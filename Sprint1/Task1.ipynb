### Структура и Объявление Класса в C#

В C#, класс объявляется с использованием ключевого слова `class`, за которым следует имя класса. Класс может содержать поля (данные), свойства (способ доступа к данным), методы (функциональность) и события. 

#### Свойства

Свойства в C# используются для инкапсуляции данных. Они предоставляют гибкий механизм для чтения, записи или вычисления значений приватных полей. Свойства могут иметь два акцессора: `get` (геттер) для чтения значения и `set` (сеттер) для его изменения.

#### Геттеры и Сеттеры

Геттеры и сеттеры являются частью свойств класса в C#. Геттер (`get`) используется для возврата значения свойства, а сеттер (`set`) — для установки нового значения. Сеттер может включать логику валидации или другие операции перед изменением значения поля.

### Пример Классов Vehicle и Car

Давайте рассмотрим примеры классов `Vehicle` и `Car`, чтобы проиллюстрировать использование свойств, геттеров и сеттеров, а также разницу между статическими и обычными атрибутами и методами.

#### Класс Vehicle

```csharp
public class Vehicle
{
    private string _color;
    private int _speed;

    // Свойство Color с геттером и сеттером
    public string Color
    {
        get { return _color; }
        set { _color = value; }
    }

    // Свойство Speed с геттером и сеттером
    public int Speed
    {
        get { return _speed; }
        set
        {
            if (value >= 0)
                _speed = value;
            else
                throw new ArgumentOutOfRangeException("Скорость не может быть отрицательной!");
        }
    }

    // Статический атрибут
    public static int TotalVehicles { get; private set; } = 0;

    // Конструктор
    public Vehicle()
    {
        TotalVehicles++;
    }

    // Обычный метод
    public void Accelerate(int delta)
    {
        Speed += delta;
    }

    // Статический метод
    public static void DisplayTotalVehicles()
    {
        Console.WriteLine($"Общее количество транспортных средств: {TotalVehicles}");
    }
}
```

#### Класс Car

```csharp
public class Car : Vehicle

    {
        FuelLevel += amount;
    }
}
```

### Разница между Статическими и Обычными Атрибутами и Методами Класса

- **Статические атрибуты и методы** принадлежат самому классу, а не его экземплярам. Они могут быть вызваны без создания объекта класса. В примере выше `TotalVehicles` является статическим атрибутом, а `DisplayTotalVehicles()` — статическим методом класса `Vehicle`.
- **Обычные атрибуты и методы** принадлежат экземплярам класса и могут иметь разные значения для разных объектов. В примере выше `_color`, `_speed` и `_fuelLevel` являются обычными атрибутами, а методы `Accelerate()` и `Refuel()` — обычными методами.

````

using System;

// Интерфейс, как в теории
public interface IVehicle
{
    // Метод ускорения, который должен быть реализован
    void Accelerate(int delta);
}

// Базовый класс Vehicle
public class Vehicle : IVehicle // Vehicle также может реализовывать IVehicle
{
    private string _color;
    private int _speed;

    // Свойство Color с геттером и сеттером
    public string Color
    {
        get { return _color; }
        set { _color = value; }
    }

    // Свойство Speed с геттером и сеттером (как в вашей теории, с валидацией)
    public int Speed
    {
        get { return _speed; }
        set
        {
            if (value >= 0)
                _speed = value;
            else
                throw new ArgumentOutOfRangeException("Скорость не может быть отрицательной!");
        }
    }

    // Статический атрибут для подсчета общего количества транспортных средств
    public static int TotalVehicles { get; private set; } = 0;

    // Конструктор Vehicle
    public Vehicle(string color = "Unknown")
    {
        Color = color; // Используем сеттер свойства Color
        Speed = 0;     // Используем сеттер свойства Speed для инициализации и валидации
        TotalVehicles++;
        // Вывод сообщения о создании прямо из конструктора
        Console.WriteLine($"[Vehicle] Создано {Color} транспортное средство. Всего транспортных средств: {TotalVehicles}");
    }

    // Метод ускорения, реализующий IVehicle
    public void Accelerate(int delta)
    {
        Speed += delta; // Используем сеттер свойства Speed для изменения скорости и валидации
        Console.WriteLine($"  '{Color}' транспортное средство ускоряется на {delta} км/ч. Текущая скорость: {Speed} км/ч.");
    }

    // Виртуальный метод для отображения информации
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"  Информация Vehicle: Цвет='{Color}', Скорость={Speed} км/ч.");
    }

    // Дополнительный метод, как в вашем примере с Animal
    public void StartEngine()
    {
        Console.WriteLine($"  '{Color}' транспортное средство заводит двигатель.");
    }

    public void StopEngine()
    {
        Console.WriteLine($"  '{Color}' транспортное средство глушит двигатель.");
    }
}

// Производный класс Car, наследует от Vehicle (и автоматически реализует IVehicle через Vehicle)
public class Car : Vehicle
{
    private int _fuelLevel;

    // Свойство FuelLevel с геттером и сеттером (как в вашей теории, с валидацией)
    public int FuelLevel
    {
        get { return _fuelLevel; }
        set
        {
            if (value >= 0 && value <= 100) // Валидация: уровень топлива от 0 до 100
                _fuelLevel = value;
            else if (value < 0)
                _fuelLevel = 0; // Не даем опуститься ниже нуля
            else
                _fuelLevel = 100; // Не даем превысить 100
        }
    }

    // Конструктор Car
    public Car(string color = "Red", int initialFuel = 20) : base(color) // Вызывает конструктор базового класса Vehicle
    {
        FuelLevel = initialFuel; // Используем сеттер свойства FuelLevel для инициализации и валидации
        // Вывод сообщения о создании прямо из конструктора
        Console.WriteLine($"[Car] Создана {Color} машина. Уровень топлива: {FuelLevel}%.");
    }

    // Метод заправки
    public void Refuel(int amount)
    {
        int oldFuel = FuelLevel;
        FuelLevel += amount; // Используем сеттер свойства FuelLevel для изменения и валидации
        Console.WriteLine($"  '{Color}' машина заправлена на {amount} л. Уровень топлива изменился с {oldFuel}% на {FuelLevel}%.");
    }

    // Дополнительный Car-специфичный метод
    public void Drive(int distance)
    {
        if (FuelLevel > 0)
        {
            int fuelConsumed = (distance / 10); // Простая симуляция расхода: 1% топлива на каждые 10 км
            if (fuelConsumed == 0 && distance > 0) fuelConsumed = 1; // Минимум 1% за любую поездку, если едет

            if (FuelLevel - fuelConsumed < 0)
            {
                Console.WriteLine($"  '{Color}' машина не может проехать {distance} км, недостаточно топлива ({FuelLevel}% осталось).");
                return;
            }

            FuelLevel -= fuelConsumed;
            Console.WriteLine($"  '{Color}' машина проехала {distance} км. Остаток топлива: {FuelLevel}%.");
        }
        else
        {
            Console.WriteLine($"  '{Color}' машина не может ехать, нет топлива!");
        }
    }

    // Переопределение метода DisplayInfo
    public override void DisplayInfo()
    {
        base.DisplayInfo(); // Вызываем метод базового класса
        Console.WriteLine($"  Информация Car: Уровень топлива={FuelLevel}%.");
    }
}

// Основная программа для демонстрации


    
    
        Console.WriteLine("--- Демонстрация классов Vehicle и Car ---");
        Console.WriteLine(); // Пустая строка для разделения

        // --- Блок 1: Создание и взаимодействие с Vehicle ---
        Vehicle myVehicle = new Vehicle("Orange"); // Вывод из конструктора Vehicle
        myVehicle.StartEngine();
        myVehicle.Accelerate(50);
        myVehicle.DisplayInfo();
        myVehicle.StopEngine();
        Console.WriteLine();

        // --- Блок 2: Создание и взаимодействие с Car ---
        Car myCar = new Car("Blue", 40); // Вывод из конструкторов Vehicle и Car
        myCar.StartEngine();
        myCar.Accelerate(70);
        myCar.Refuel(10);
        Console.WriteLine($"  Текущий уровень топлива '{myCar.Color}': {myCar.FuelLevel}%."); // Небольшое уточнение
        myCar.Drive(150);
        myCar.DisplayInfo();
        myCar.StopEngine();
        Console.WriteLine();

        // --- Блок 3: Демонстрация полиморфизма ---
        // Создаем объект Car, но ссылаемся на него через базовый класс Vehicle
        Vehicle polymorphicCar = new Car("Green", 30); // Вывод из конструкторов Vehicle и Car
        polymorphicCar.StartEngine();
        polymorphicCar.Accelerate(90); // Вызываем метод Accelerate базового класса
        polymorphicCar.DisplayInfo();  // Вызываем переопределенный метод DisplayInfo из Car
        // Чтобы вызвать Car-специфичный метод, нужно привести тип:
        Console.WriteLine($"  Полиморфная машина (доступ к специфичным методам Car):");
        ((Car)polymorphicCar).Refuel(5);
        ((Car)polymorphicCar).Drive(100);
        polymorphicCar.StopEngine();
        Console.WriteLine();

        // --- Блок 4: Демонстрация валидации ---
        Console.WriteLine("--- Демонстрация валидации свойств ---");
        try
        {
            myVehicle.Speed = -10; // Попытка установить отрицательную скорость
        }
        catch (ArgumentOutOfRangeException ex)
        {
            Console.WriteLine($"  Ошибка установки скорости: {ex.Message}");
        }
        Console.WriteLine($"  Скорость '{myVehicle.Color}' после попытки: {myVehicle.Speed} км/ч (осталась прежней).");

        Console.WriteLine($"  Уровень топлива '{myCar.Color}' до попытки: {myCar.FuelLevel}%.");
        myCar.FuelLevel = 120; // Попытка установить слишком высокий уровень
        Console.WriteLine($"  Уровень топлива '{myCar.Color}' после попытки 120%: {myCar.FuelLevel}% (должен быть 100).");
        myCar.FuelLevel = -5;  // Попытка установить слишком низкий уровень
        Console.WriteLine($"  Уровень топлива '{myCar.Color}' после попытки -5%: {myCar.FuelLevel}% (должен быть 0).");
        Console.WriteLine();

        // --- Итоговый вывод ---
        Console.WriteLine($"Всего транспортных средств создано: {Vehicle.TotalVehicles}");

        
    
````

````
using System;

// --- Базовый класс Animal ---
public class Animal
{
    private string _name;
    private int _age;
    private string _species; // Вид животного, например, "Лев", "Собака", "Кошка"

    // Свойство Name с геттером и сеттером
    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    // Свойство Age с геттером и сеттером (с валидацией, как в теории)
    public int Age
    {
        get { return _age; }
        set
        {
            if (value >= 0)
                _age = value;
            else
                throw new ArgumentOutOfRangeException("Возраст не может быть отрицательным!");
        }
    }

    // Свойство Species (только геттер, устанавливается в конструкторе)
    public string Species
    {
        get { return _species; }
        private set { _species = value; }
    }

    // Статический атрибут для подсчета общего количества животных
    public static int TotalAnimals { get; private set; } = 0;

    // Конструктор Animal
    public Animal(string species, string name, int age)
    {
        Species = species;
        Name = name;
        Age = age; // Используем сеттер для валидации возраста
        TotalAnimals++;
        // Вывод сообщения о создании прямо из конструктора
        Console.WriteLine($"[Animal] Создан {Species} по имени {Name}. Возраст: {Age}. Всего животных: {TotalAnimals}");
    }

    // Обычные (экземплярные) методы
    public void Eat()
    {
        Console.WriteLine($"{Name} ест.");
    }

    public void Move()
    {
        Console.WriteLine($"{Name} двигается.");
    }

    public void Sleep()
    {
        Console.WriteLine($"{Name} спит.");
    }

    public void WakeUp()
    {
        Console.WriteLine($"{Name} просыпается.");
    }

    // Виртуальный метод для отображения общей информации (можно переопределить)
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"  Общая информация: Тип={Species}, Имя={Name}, Возраст={Age}.");
    }

    // Статический метод для вывода общего количества животных
    public static void DisplayTotalAnimalsCount()
    {
        Console.WriteLine($"\nВсего животных создано: {TotalAnimals}");
    }
}

// --- Производный класс Dog ---
public class Dog : Animal
{
    private string _breed;
    private bool _isTrained;

    // Свойство Breed с геттером и сеттером
    public string Breed
    {
        get { return _breed; }
        set { _breed = value; }
    }

    // Свойство IsTrained с геттером и сеттером
    public bool IsTrained
    {
        get { return _isTrained; }
        set { _isTrained = value; }
    }

    // Конструктор Dog
    public Dog(string name, int age, string breed, bool isTrained = true)
        : base("Собака", name, age) // Вызываем конструктор базового класса Animal
    {
        Breed = breed;
        IsTrained = isTrained;
        Console.WriteLine($"[Dog] Это собака породы {Breed}. Обученность: {(IsTrained ? "обучена" : "не обучена")}.");
    }

    // Специфичные для Dog методы
    public void Bark()
    {
        Console.WriteLine($"{Name} лает: Гав-гав!");
    }

    public void Fetch()
    {
        Console.WriteLine($"{Name} приносит мячик.");
    }

    public void WagTail()
    {
        Console.WriteLine($"{Name} бегает и виляет хвостом!");
    }

    // Переопределение метода DisplayInfo
    public override void DisplayInfo()
    {
        base.DisplayInfo(); // Вызываем метод базового класса
        Console.WriteLine($"  Информация о Dog: Порода={Breed}, Обучена={IsTrained}.");
    }
}

// --- Производный класс Cat ---
public class Cat : Animal
{
    private bool _hasClaws; // Наличие когтей
    private string _furColor; // Цвет шерсти

    // Свойство HasClaws с геттером и сеттером
    public bool HasClaws
    {
        get { return _hasClaws; }
        set { _hasClaws = value; }
    }

    // Свойство FurColor с геттером и сеттером
    public string FurColor
    {
        get { return _furColor; }
        set { _furColor = value; }
    }

    // Конструктор Cat
    public Cat(string name, int age, string furColor, bool hasClaws = true)
        : base("Кошка", name, age) // Вызываем конструктор базового класса Animal
    {
        FurColor = furColor;
        HasClaws = hasClaws;
        Console.WriteLine($"[Cat] Это кошка по имени {Name}. Когти: {(HasClaws ? "есть" : "нет")}, Цвет шерсти: {FurColor}.");
    }

    // Специфичные для Cat методы
    public void Meow()
    {
        Console.WriteLine($"{Name} мяукает: Мяу!");
    }

    public void Climb()
    {
        Console.WriteLine($"{Name} ловко забирается на дерево.");
    }

    public void Purr()
    {
        Console.WriteLine($"{Name} мурлычет.");
    }

    // Переопределение метода DisplayInfo
    public override void DisplayInfo()
    {
        base.DisplayInfo(); // Вызываем метод базового класса
        Console.WriteLine($"  Информация о Cat: Цвет шерсти={FurColor}, Когти={(HasClaws ? "есть" : "нет")}.");
    }
}


// --- Основная программа для демонстрации ---


    
        Console.WriteLine("--- Демонстрация класса Animal и его производных ---");
        Console.WriteLine();

        // --- Блок 1: Создание и взаимодействие с Animal (Лев Симба) ---
        // Используем базовый класс Animal, чтобы создать "Льва"
        Animal lion = new Animal("Лев", "Симба", 5); // Сообщение из конструктора Animal
        lion.Eat();
        lion.Move();
        lion.Sleep();
        lion.WakeUp();
        lion.DisplayInfo();
        Console.WriteLine();

        // --- Блок 2: Создание и взаимодействие с Dog (Рекс) ---
        Dog rex = new Dog("Рекс", 3, "Немецкая овчарка"); // Сообщения из конструкторов Animal и Dog
        rex.Bark();
        rex.WagTail();
        rex.Fetch();
        rex.DisplayInfo();
        Console.WriteLine();

        // --- Блок 3: Создание и взаимодействие с Cat (Усик) ---
        Cat usik = new Cat("Усик", 2, "Рыжий", true); // Сообщения из конструкторов Animal и Cat
        usik.Meow();
        usik.Climb();
        usik.Purr();
        usik.DisplayInfo();
        Console.WriteLine();

        // --- Демонстрация валидации возраста ---
        Console.WriteLine("--- Демонстрация валидации возраста ---");
        Console.WriteLine($"Текущий возраст {lion.Name}: {lion.Age}.");
        try
        {
            lion.Age = -1; // Попытка установить недопустимый возраст
        }
        catch (ArgumentOutOfRangeException ex)
        {
            Console.WriteLine($"  Ошибка: {ex.Message}");
        }
        Console.WriteLine($"Возраст {lion.Name} после попытки: {lion.Age} (должен остаться прежним).");
        Console.WriteLine();

        // --- Итоговый вывод ---
        Animal.DisplayTotalAnimalsCount(); // Вызываем статический метод для общего подсчета

        
    

````
