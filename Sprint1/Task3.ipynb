различные опции доставки. На основе этого класса разработать 2-3 производных 
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из

Требования к базовому классу ShippingOption: 
• Атрибуты: ID опции доставки (DeliveryOptionId), Название опции доставки 
(DeliveryOptionName), Стоимость доставки (Cost). 
• Методы: 
o CalculateCost(): метод для расчета стоимости доставки. 
o EstimateDeliveryTime(): метод для оценки времени доставки.

```
public class ShippingOption
{
    // Атрибуты с использованием get; set; (автоматические свойства)
    // Эти свойства обеспечивают геттеры и сеттеры для атрибутов.
    public int DeliveryOptionId { get; set; }
    public string DeliveryOptionName { get; set; }
    public decimal Cost { get; set; }

    // Конструктор класса с использованием геттеров и сеттеров (через свойства)
    public ShippingOption(int id, string name, decimal cost)
    {
        DeliveryOptionId = id;      // Использует сеттер для DeliveryOptionId
        DeliveryOptionName = name;  // Использует сеттер для DeliveryOptionName
        Cost = cost;                // Использует сеттер для Cost
    }

    // Виртуальный метод для расчета стоимости доставки.
    // Может быть переопределен в производных классах.
    public virtual decimal CalculateCost()
    {
        return Cost;
    }

    // Виртуальный метод для оценки времени доставки.
    // Может быть переопределен в производных классах.
    public virtual string EstimateDeliveryTime()
    {
        return "Время доставки будет оценено.";
    }

    // Виртуальный метод для получения деталей опции доставки.
    // Может быть переопределен в производных классах.
    public virtual string GetDeliveryDetails()
    {
        return $"ID: {DeliveryOptionId}, Название: {DeliveryOptionName}, Базовая стоимость: {Cost:C}";
    }
}

public class StandardDelivery : ShippingOption
{
    // Дополнительный атрибут для средней продолжительности доставки (в днях)
    public int AverageDeliveryTimeDays { get; set; }

    // Конструктор с вызовом базового конструктора и инициализацией нового атрибута
    public StandardDelivery(int id, string name, decimal cost, int averageTimeDays)
        : base(id, name, cost) // Вызов конструктора базового класса
    {
        AverageDeliveryTimeDays = averageTimeDays; // Использование сеттера для AverageDeliveryTimeDays
    }

    // Переопределение метода EstimateDeliveryTime() для отображения среднего времени
    public override string EstimateDeliveryTime()
    {
        return $"Примерное время доставки: {AverageDeliveryTimeDays} дней (стандартная).";
    }

    // В данном классе CalculateCost и GetDeliveryDetails не переопределяются
    // и используют реализации из базового класса.
}

public class ExpressDelivery : ShippingOption
{
    // Дополнительный атрибут для минимального времени доставки (в часах)
    public int MinDeliveryTimeHours { get; set; }
    private decimal _expressSurcharge = 10.0m; // Примерная наценка за экспресс-доставку

    // Конструктор с вызовом базового конструктора и инициализацией нового атрибута
    public ExpressDelivery(int id, string name, decimal cost, int minTimeHours)
        : base(id, name, cost) // Вызов конструктора базового класса
    {
        MinDeliveryTimeHours = minTimeHours; // Использование сеттера для MinDeliveryTimeHours
    }

    // Переопределение метода CalculateCost() для увеличения стоимости
    public override decimal CalculateCost()
    {
        // Увеличиваем базовую стоимость на наценку
        return base.CalculateCost() + _expressSurcharge;
    }

    // Переопределение метода EstimateDeliveryTime() для отображения минимального времени
    public override string EstimateDeliveryTime()
    {
        return $"Примерное время доставки: {MinDeliveryTimeHours} часов (экспресс).";
    }

    // Переопределение метода GetDeliveryDetails() для включения информации об экспресс-наценке
    public override string GetDeliveryDetails()
    {
        return base.GetDeliveryDetails() + $", Экспресс-наценка: {_expressSurcharge:C}, Минимальное время: {MinDeliveryTimeHours} ч.";
    }
}

public class Pickup : ShippingOption
{
    // Дополнительный атрибут для адреса пункта самовывоза
    public string PickupAddress { get; set; }

    // Конструктор с вызовом базового конструктора и инициализацией нового атрибута
    public Pickup(int id, string name, decimal cost, string pickupAddress)
        : base(id, name, cost) // Вызов конструктора базового класса
    {
        PickupAddress = pickupAddress; // Использование сеттера для PickupAddress
    }

    // Переопределение метода CalculateCost() (для самовывоза часто 0 или номинальная стоимость)
    public override decimal CalculateCost()
    {
        // Стоимость самовывоза берется из базовой стоимости, которая может быть 0
        return Cost;
    }

    // Переопределение метода EstimateDeliveryTime() для самовывоза
    public override string EstimateDeliveryTime()
    {
        return "Заказ готов к самовывозу (обычно в течение 1-2 часов после оформления).";
    }

    // Переопределение метода GetDeliveryDetails() для отображения адреса пункта самовывоза
    public override string GetDeliveryDetails()
    {
        return base.GetDeliveryDetails() + $", Адрес пункта самовывоза: {PickupAddress}";
    }
}


        // Установка кодировки для корректного отображения кириллицы в консоли
        Console.OutputEncoding = System.Text.Encoding.UTF8;

        // Создаем список базового типа ShippingOption для демонстрации полиморфизма.
        // Это позволяет хранить объекты всех производных классов.
        List<ShippingOption> deliveryOptions = new List<ShippingOption>();

        // Создаем экземпляры различных опций доставки, используя их конструкторы
        deliveryOptions.Add(new StandardDelivery(101, "Стандартная доставка", 5.00m, 3));
        deliveryOptions.Add(new ExpressDelivery(102, "Экспресс доставка (Срочная)", 15.00m, 24)); // Базовая стоимость 15, добавится наценка
        deliveryOptions.Add(new Pickup(103, "Самовывоз из пункта выдачи", 0.00m, "г. Москва, ул. Примерная, д. 15, офис 201"));
        deliveryOptions.Add(new StandardDelivery(104, "Доставка по городу (Эконом)", 7.50m, 2));
        deliveryOptions.Add(new ExpressDelivery(105, "Экспресс доставка (Супер-срочная)", 25.00m, 12));
        deliveryOptions.Add(new Pickup(106, "Самовывоз со склада", 2.50m, "г. Санкт-Петербург, пр. Заводской, д. 50"));


        Console.WriteLine("--- Демонстрация различных опций доставки и взаимодействия объектов ---");
        Console.WriteLine("====================================================================");

        // Итерируем по списку объектов.
        // Благодаря полиморфизму, будут вызываться переопределенные методы
        // соответствующего производного класса. Это и есть взаимодействие объектов.
        foreach (var option in deliveryOptions)
        {
            Console.WriteLine($"\n*** Тип опции: {option.GetType().Name} ***"); // Выводит фактический тип объекта (StandardDelivery, ExpressDelivery, Pickup)
            Console.WriteLine(option.GetDeliveryDetails());      // Вызов переопределенного GetDeliveryDetails
            Console.WriteLine($"Расчетная стоимость: {option.CalculateCost():C}"); // Вызов переопределенного CalculateCost
            Console.WriteLine(option.EstimateDeliveryTime());    // Вызов переопределенного EstimateDeliveryTime
            Console.WriteLine("--------------------------------------------------------------------");
        }

        Console.WriteLine("\n--- Завершение демонстрации ---");
    
```
